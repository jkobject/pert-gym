```python
import anndata as ad
import mudata as mu
import tempfile
from pathlib import Path

# the 4 Papalexi files from `unprocessed`
papalexi_paths = unprocessed[1:5]

# load as AnnData from Lamin artifacts
papalexi = {p: ln.Artifact.filter(key__contains=p).one().load() for p in papalexi_paths}

# concat by modality
rna = ad.concat(
    [papalexi[papalexi_paths[0]], papalexi[papalexi_paths[3]]],
    join="outer",
    label="source",
    keys=["eccite_arrayed_RNA", "eccite_RNA"],
    index_unique="-",
)
protein = ad.concat(
    [papalexi[papalexi_paths[1]], papalexi[papalexi_paths[2]]],
    join="outer",
    label="source",
    keys=["eccite_arrayed_protein", "eccite_protein"],
    index_unique="-",
)

# add protein features into RNA obs (few columns, so store as metadata)
protein_df = protein.to_df().reindex(rna.obs_names)
for col in protein_df.columns:
    rna.obs[f"protein_{col}"] = protein_df[col].values

# migrate the single AnnData with protein values in obs
try:
    migrate_h5ad_to_triplet(rna, ln, replace_on_instance=True, storage=False, dataset_prefix="papalexi_eccite_merged")
except TypeError:
    with tempfile.TemporaryDirectory() as td:
        tmp_path = Path(td) / "papalexi_eccite_merged.h5ad"
        rna.write_h5ad(tmp_path)
        migrate_h5ad_to_triplet(
            str(tmp_path), ln, replace_on_instance=True, storage=False
        )
```

```python
def _concat_and_migrate(paths, prefix):
    adatas = [ln.Artifact.filter(key__contains=p).one().load() for p in paths]
    merged = ad.concat(
        adatas,
        join="outer",
        label="source",
        keys=[Path(p).stem for p in paths],
        index_unique="-",
    )
    try:
        migrate_h5ad_to_triplet(merged, ln, replace_on_instance=True, storage=False, dataset_prefix=prefix)
    except TypeError:
        with tempfile.TemporaryDirectory() as td:
            tmp_path = Path(td) / f"{prefix}.h5ad"
            merged.write_h5ad(tmp_path)
            migrate_h5ad_to_triplet(
                str(tmp_path), ln, replace_on_instance=True, storage=False
            )


# GSE106340, GSE115943: concat then migrate
_concat_and_migrate(unprocessed[6:8], prefix="SchiebingerLander2019")

# GSE120861: concat then migrate
_concat_and_migrate(unprocessed[8:11], prefix="GasperiniShendure2019")
```
